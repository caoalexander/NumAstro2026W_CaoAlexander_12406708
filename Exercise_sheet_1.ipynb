## Exercise 1 - FizzBuzz

Write a program that prints integer numbers from 1 to 25. For multiples of 3 print `"Fizz"` instead of the number, and for multiples of 5 print `"Buzz"`.

**Careful:** for multiples of 3 and 5, you should print `"FizzBuzz"`!

```python
# Loop through the numbers from 1 to 25
for i in range(1, 26):

    # Check if the number fulfills the conditions
    if i % 3 == 0:

        if i % 5 == 0:
            print("FizzBuzz")
        else:
            print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)


```

## Exercise 2 - Leap years

**(a)** Write a function that, given a year, reports if it is a [leap year](https://en.wikipedia.org/wiki/Leap_year). The tricky thing is that a leap year in the Gregorian calendar occurs:
 * on every year that is evenly divisible by 4
 * except every year that is evenly divisible by 100
 * unless the year is also evenly divisible by 400
 
For example, 1995 is not a leap year, but 1996 is. 1900 is not a leap year, but 2000 is.

**(b)** Test your function on some selected years that allow you to check if it provides the correct solution.

```python
import random

# Generate a random year between 0 and 3000
year = random.randint(0, 3000)

# Check if the year fulfills the conditions
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")

```

## Exercise 4 - Find the fake coin

Consider this famous puzzle: _"You have 8 coins: all the coins have equal weight, except one. The odd one weights less than the others, not being of pure gold. In the worst case, how many iterations do you need to find the odd one out on a two plates scale?"_

Write a function that returns the minimum amount of weighing it will take to measure `N` coins in every possible occurrence (including worst case scenario, i.e.: without relying on luck at all!). The number `N` is guaranteed to be a positive integer.

**Tip:** being able to think recursively might help here!

**Note:** this is clearly a logical problem, because a good coder is a master of pattern recognition and subsequent optimization! ;)

```python
import numpy as np

# Generate a random number of coins between 1 and 100
coin_num = random.randint(1, 100)

# Define a function to calculate the number of steps needed
# to find a coin, based on the number of coins being examined
def find_coin(coin_num):

    # If there is only one coin or fewer, no further steps are needed
    if coin_num <= 1:
        return [0]

    # Divide the coins into three groups as evenly as possible
    # If the number is not divisible by 3, the groups may differ in size
    if coin_num % 3 != 0: 
        group_sizes = [(coin_num + 2) // 3, coin_num // 3] 
    else: 
        group_sizes = [coin_num // 3]

    # Create an empty list to store the results
    counter = []

    # Go through each group size
    for size in group_sizes:

        # Recursively calculate the steps for each group
        # Add 1 to account for the current step
        # Extend adds all the returned values to the counter list
        counter.extend([1 + i for i in find_coin(size)])

    return counter

counter = np.sort(find_coin(coin_num))
print(counter)

print(f"Number of coins: {coin_num}")

i = 0
for amount in set(counter):
    i = i + 1
    if amount == np.max(counter):
        print(f"Worst case scenario: {amount}")
    else:
        print(f"{i}. minimum: {amount}")

```
